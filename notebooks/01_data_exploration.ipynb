Objectif : explorer les premières données de marché Yahoo Finance et valider les champs attendus avant ingestion BigQuery.

```python
import pandas as pd
import numpy as np
import yfinance as yf
```

## Exemple de récupération

Remplacer le ticker ci-dessous par un instrument validé lors du cadrage.

```python
ticker = "GC=F"  # Exemple : futures or
start_date = "2020-01-01"

data = yf.download(ticker, start=start_date, progress=False)
data.head()
```

## Contrôles rapides

- vérifier la couverture temporelle ;
- vérifier les valeurs nulles ;
- vérifier les colonnes OHLCV ;
- identifier les dates manquantes ou doublons.

```python
summary = {
    "rows": len(data),
    "start": data.index.min(),
    "end": data.index.max(),
    "nulls": data.isna().sum().to_dict(),
}
summary
```
